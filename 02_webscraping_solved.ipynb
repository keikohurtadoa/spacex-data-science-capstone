  ## Objectives
Web scrap Falcon 9 launch records with `BeautifulSoup`: 
- Extract a Falcon 9 launch records HTML table from Wikipedia
- Parse the table and convert it into a Pandas data frame


First let's import required packages for this lab


````python
# If needed:
# %pip install beautifulsoup4 requests pandas
````

````python
import sys

import requests
from bs4 import BeautifulSoup
import re
import unicodedata
import pandas as pd
````

and we will provide some helper functions for you to process web scraped HTML table


````python
def clean_text(cell):
    copy = BeautifulSoup(str(cell), 'html.parser')
    for ref in copy.select('sup.reference'):
        ref.decompose()
    return unicodedata.normalize('NFKC', copy.get_text(' ', strip=True))

def date_time(cell):
    text = clean_text(cell)
    date_match = re.search(r'\d{1,2}\s+[A-Za-z]+\s+\d{4}', text)
    time_match = re.search(r'\b\d{2}:\d{2}(?::\d{2})?\b', text)
    return [date_match.group() if date_match else text,
            time_match.group() if time_match else None]

def booster_version(cell):
    return clean_text(cell)

def landing_status(cell):
    return clean_text(cell)

def get_mass(cell):
    text = clean_text(cell)
    match = re.search(r'[\d,]+(?:\.\d+)?\s*kg', text)
    return match.group() if match else text

def extract_column_from_header(cell):
    text = clean_text(cell)
    return text if text and not text.isdigit() else None
````

### TASK 2: Extract all column/variable names from the HTML table header


Next, we want to collect all relevant column names from the HTML table header


Let's try to find all tables on the wiki page first. If you need to refresh your memory about `BeautifulSoup`, please check the external reference link towards the end of this lab


````python
html_tables = soup.find_all('table')
````

Starting from the third table is our target table contains the actual launch records.


````python
# Identify a launch table by its header rather than a brittle table index.
launch_tables = [table for table in soup.select('table.wikitable')
                 if 'Flight No.' in table.get_text(' ', strip=True)
                 and 'Payload mass' in table.get_text(' ', strip=True)]
if not launch_tables:
    raise ValueError('No launch tables found. Check that the historical Wikipedia page loaded.')
first_launch_table = launch_tables[0]
print(first_launch_table.find('tr'))
````

Next, we just need to iterate through the `<th>` elements and apply the provided `extract_column_from_header()` to extract column name one by one


````python
column_names = []
for th in first_launch_table.find('tr').find_all('th'):
    name = extract_column_from_header(th)
    if name is not None and len(name) > 0:
        column_names.append(name)
````

## TASK 3: Create a data frame by parsing the launch HTML tables


We will create an empty dictionary with keys from the extracted column names in the previous task. Later, this dictionary will be converted into a Pandas dataframe


````python
launch_dict = {name: [] for name in [
    'Flight No.', 'Date', 'Time', 'Version Booster', 'Launch site',
    'Payload', 'Payload mass', 'Orbit', 'Customer', 'Launch outcome', 'Booster landing'
]}
````

Next, we just need to fill up the `launch_dict` with launch records extracted from table rows.


Usually, HTML tables in Wiki pages are likely to contain unexpected annotations and other types of noises, such as reference links `B0004.1[8]`, missing values `N/A [e]`, inconsistent formatting, etc.


To simplify the parsing process, we have provided an incomplete code snippet below to help you to fill up the `launch_dict`. Please complete the following code snippet with TODOs or you can choose to write your own logic to parse all launch tables:


````python
extracted_row = 0
for table in launch_tables:
    for row in table.find_all('tr'):
        heading = row.find('th', recursive=False)
        cells = row.find_all('td', recursive=False)
        if heading is None or len(cells) != 9:
            continue  # Skip header, commentary and merged-cell rows.
        # Strip references before identifying the flight number.
        flight_number = clean_text(heading)
        if not flight_number.isdigit():
            continue
        date, time = date_time(cells[0])
        values = [int(flight_number), date, time, booster_version(cells[1]),
                  clean_text(cells[2]), clean_text(cells[3]), get_mass(cells[4]),
                  clean_text(cells[5]), clean_text(cells[6]), clean_text(cells[7]),
                  landing_status(cells[8])]
        for key, value in zip(launch_dict, values):
            launch_dict[key].append(value)
        extracted_row += 1
assert extracted_row > 0, 'No launch records extracted.'
print('Launch records extracted:', extracted_row)
````

After you have fill in the parsed launch record values into `launch_dict`, you can create a dataframe from it.


````python
df = pd.DataFrame(launch_dict)
assert df['Flight No.'].is_unique
df.head()
````
